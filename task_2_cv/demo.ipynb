### Seasonal Image Matching - Demo

This notebook showcases the pipeline for matching seasonal satellite image patches

```python
import os
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

plt.style.use('seaborn-v0_8-darkgrid')
plt.rcParams['figure.figsize'] = (15, 8)
```

```python
IMG1 = "data/processed/val/winter/patch_0011.png"
IMG2 = "data/processed/val/summer/patch_0011.png"
OUTPUT_FILE = "results/demo_output.png"

!python inference.py --img1 {IMG1} --img2 {IMG2} --output {OUTPUT_FILE}
```

```python
def show_result(output_path):
    if not os.path.exists(output_path):
        print(f"❌ Зображення не знайдено за шляхом: {output_path}")
        return
    
    img = mpimg.imread(output_path)
    fig, ax = plt.subplots()
    ax.imshow(img)
    ax.axis('off')
    ax.set_title("Inference Result: Inliers (Green) & Outliers (Red)", fontsize=14)
    plt.show()

show_result(OUTPUT_FILE)
```
