## Mountain Named Entity Recognition (NER)
This notebook demonstrates the operation of a model based on the Transformer architecture  that has been fine-tuned to recognize mountain names in texts.

```python
import os
import re
from IPython.display import display, HTML
from inference import MountainPredictor

predictor = MountainPredictor(model_path="./saved_model")
```

## Creating Test text:

```python
sample_text = """We started our journey in the Himalayas.
Our main goal was to see Mount Everest with our own eyes, but the weather was tough.
Many climbers agree that K2 is much more dangerous to climb.
After a week, we returned to Europe and decided to visit Mont Blanc.
The views there are completely different from Mount Kilimanjaro in Africa.
Tomorrow we are just going to relax near the local lake.
During our expedition, we climbed Mount Everest, took a break, and then headed towards K2 and Mont Blanc.
Yesterday I went to the supermarket, bought some coffee, and read a book in Central Park.
Is Kangchenjunga steeper than Annapurna or Mount Elbrus?"""

file_path = "mountains_text.txt"

with open(file_path, "w", encoding="utf-8") as f:
    f.write(sample_text)
    
```

## Results  Visualize:

```python
with open(file_path, "r", encoding="utf-8") as f:
    lines = [line.strip() for line in f.readlines() if line.strip()]
    
def visualize_ner(predict_result):
    mountains = predict_result['extracted_mountains']
    text = predict_result['text']
    
    mountains = sorted(list(set(mountains)), key=len, reverse=True)
    
    for m in mountains:
        if not m.strip():
            continue
            
        pattern = re.compile(re.escape(m), re.IGNORECASE)
        
        text = pattern.sub(
            lambda match: f"<span style='font-weight: bold; padding: 2px 4px; border-radius: 4px; color: #cc5500;'>{match.group(0)}</span>",
            text
        )
        
    display(HTML(f"<div style='font-size: 16px; line-height: 1.5;'>{text}</div>"))

for line in lines:
    res = predictor.predict(line)
    visualize_ner(res)
    print()
```
